```python
from rdkit.Chem import AllChem as Chem
from rdkit.Chem.PandasTools import RenderImagesInAllDataFrames

RenderImagesInAllDataFrames()

from gloryxr import GLORYxR
from gloryxr.utils import reactions_to_table

gloryxr = GLORYxR()
```

```python
mol = Chem.MolFromSmiles("CCCO")
```

```python
reactions = gloryxr.predict_one(mol)
```

```python
reactions_to_table([rxn.concrete_reaction for rxn in reactions]).assign(
    Score=[rxn.score for rxn in reactions]
)
```
