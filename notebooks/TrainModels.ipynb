```python
import subprocess
from pathlib import Path

from tqdm.auto import tqdm
```

```python
OUTPUT_PATH = Path("../models")
```

```python
model_paths = {
    "Phase 1 SyGMa rules": Path("../datasets/MetaQSAR_data/p1.sdf"),
    "CYP rules from GLORY (phase 1)": Path("../datasets/MetaQSAR_data/p1.sdf"),
    "Other phase 2 rules": Path("../datasets/MetaQSAR_data/p2.sdf"),
    "UGT rules (phase 2)": Path("../datasets/MetaQSAR_data/GLORYxR_specials/UGT.sdf"),
    "GST rules (phase 2)": Path("../datasets/MetaQSAR_data/GLORYxR_specials/GST.sdf"),
    "NAT rules (phase 2)": Path("../datasets/MetaQSAR_data/GLORYxR_specials/NAT.sdf"),
    "SULT rules (phase 2)": Path("../datasets/MetaQSAR_data/GLORYxR_specials/SULT.sdf"),
    "MT rules (phase 2)": Path("../datasets/MetaQSAR_data/GLORYxR_specials/MT.sdf"),
}
```

```python
for name, path in tqdm(model_paths.items()):
    subprocess.check_output(
        [
            "fame3r",
            "train",
            "-i",
            path,
            "-o",
            OUTPUT_PATH,
            "--radius",
            "5",
        ]
    )

    (OUTPUT_PATH / "random_forest_classifier.joblib").rename(
        OUTPUT_PATH / f"{name}.joblib"
    )
```
